```python
# Datos de prueba
data = [
    (1, "kevin", 1000),
    (2, "mara", 1100),
    (3, "jose", 1300)
]

columns = ["id", "name", "salary"]

# Crear DataFrame en Spark
df = spark.createDataFrame(data, columns)

# Guardar como tabla Delta en tu schema (con backticks)
df.write.format("delta").saveAsTable("workspace.`meli-price_intelligence`.smoke_test")

# Leer la tabla Delta (con backticks)
df_read = spark.table("workspace.`meli-price_intelligence`.smoke_test")

# Mostrar los datos
df_read.show()
```
