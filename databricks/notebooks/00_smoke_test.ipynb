```python
# Prueba manual del entorno GTFS: los defaults describen el entorno actual.
# No se ejecuta como parte del Job Bronze.
dbutils.widgets.text("catalog", "workspace", "Catálogo")
dbutils.widgets.text("schema", "gtfs_bogota", "Schema")
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()
smoke_table = f"`{catalog}`.`{schema}`.`smoke_test`"

# Datos de prueba
data = [
    (1, "kevin", 1000),
    (2, "mara", 1100),
    (3, "jose", 1300)
]

columns = ["id", "name", "salary"]

# Crear DataFrame en Spark
df = spark.createDataFrame(data, columns)

# Guardar y leer la prueba en el schema GTFS seleccionado.
df.write.format("delta").saveAsTable(smoke_table)

# Verificar que la tabla se puede consultar.
df_read = spark.table(smoke_table)

# Mostrar los datos
df_read.show()
```
